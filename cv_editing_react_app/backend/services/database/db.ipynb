# Analyze the Job Data

```python
from sqlalchemy import create_engine, delete
from sqlalchemy.orm import Session
from pathlib import Path
from db_schema import Job
from db import init_db, add_a_job, get_all_jobs,remove_db, delete_all_jobs,delete_job_by_id, get_job_id, dict_to_job, job_to_dict



db_path = Path.cwd().parents[1] / "jobs.db"
DATABASE_URL = f"sqlite:///{db_path}"
engine = create_engine(DATABASE_URL)

# from db_schema import Base
# Base.metadata.create_all(engine)
```

```python
import json

with open('/home/mohan.dash/AI_CV_Customization/job_listings/job_listings_20260619_132819.json', 'r') as f:
    jobs = json.load(f)

job_data = jobs[1]

job = dict_to_job(job_data)

```

```python


job_id = job.id
with Session(engine) as session:
    existing_job = session.get(Job, job_id)
    if existing_job is None:
        session.add(job)
        session.commit()
    else:
        print("Job already exists in the database")
```

```python
with Session(engine) as session:
    jobs = session.query(Job).all()
    
job_dict = [job_to_dict(j) for j in jobs]
```

```python
from sqlalchemy import text


with engine.connect() as conn:
    conn.execute(text("ALTER TABLE jobs DROP COLUMN recruiter_name"))
    conn.execute(text("ALTER TABLE jobs DROP COLUMN recruiter_profile"))
    conn.execute(text("ALTER TABLE jobs DROP COLUMN recruiter_title"))
    conn.commit()
```

```python
from collections import Counter
all_keywords = []
for j in job_dict:
    all_keywords.extend(j["job_keywords"])
all_keywords = [k.lower() for k in all_keywords]
Counter(all_keywords).most_common(40)
```

```python
from collections import Counter
all_job_titles = []
for j in job_dict:
    all_job_titles.append(j["job_title"])
all_job_titles = [k.lower() for k in all_job_titles]
Counter(all_job_titles).most_common(40)
```

```python
all_companies_location = []
for j in job_dict:
    all_companies_location.append((j["company_name"], j["job_location"]))
all_companies_location = [(k.lower(), l.lower()) for k,l in all_companies_location]
```
